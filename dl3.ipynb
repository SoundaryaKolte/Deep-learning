```python
# 1. Import libraries
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, Flatten
import numpy as np

# 2. Load dataset
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)

# 3. Pad sequences (makes all reviews same length)
maxlen = 200
X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)

# 4. Build DNN model (simpler approach)
model = Sequential([
    Embedding(input_dim=10000, output_dim=32, input_length=maxlen),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])

# 5. Compile
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 6. Train
model.fit(X_train, y_train, epochs=5, batch_size=512, verbose=0)

# 7. Evaluate
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print("Test Accuracy:", accuracy)

# 8. Predictions
predictions = model.predict(X_test[:5])

print("\nSample Predictions:\n")
for i in range(5):
    pred = "Positive" if predictions[i][0] > 0.5 else "Negative"
    actual = "Positive" if y_test[i] == 1 else "Negative"
    print(f"Actual: {actual}, Predicted: {pred}")
```
